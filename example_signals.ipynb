```python
import matplotlib.pyplot as plt
from electro.dsp import signals as sig
import numpy as np
```

# Example Signals
---

## Common Values

```python
NN = 500      # cantidad de muestras
FS = 1000.0   # frecuencia de muestreo [Hz]
FF = 5.0      # frecuencia de la señal [Hz]

tt = sig.time_axis(nn=NN, fs=FS)
```

## Common Signals

```python
x_sin = sig.sin(vmax=2, dc=0.5, ff=FF, ph=0, nn=NN, fs=FS)
x_square = sig.square(vmax=1, dc=0, ff=FF, duty=0.3, nn=NN, fs=FS)
x_saw = sig.sawtooth(vmax=1, dc=0, ff=FF, nn=NN, fs=FS)
x_tri = sig.triangle(vmax=1, dc=0, ff=FF, duty=0.5, nn=NN, fs=FS)
  
# --- Gráficos ---
fig, axs = plt.subplots(4, 1, figsize=(12, 10))
 
axs[0].plot(tt, x_sin)
axs[0].set_title("Sinusoide (vmax=2, dc=0.5, 5 Hz)")
 
axs[1].plot(tt, x_square)
axs[1].set_title("Cuadrada (duty=0.3)")
 
axs[2].plot(tt, x_saw)
axs[2].set_title("Diente de sierra")
 
axs[3].plot(tt, x_tri)
axs[3].set_title("Triangular")
  
for ax in axs.flat:
    ax.set_xlabel("t [s]")
    ax.grid(True)
 
fig.tight_layout()
plt.show() 
```

## Rectificadores

```python
x_rmmo = sig.rmmo(vmax=311, ff=FF, ph=0, nn=NN, fs=FS)      # media onda monofásico
x_rmoc = sig.rmoc(vmax=311, ff=FF, ph=0, nn=NN, fs=FS)      # onda completa monofásico
x_rtmo = sig.rtmo(vmax=311, ff=FF, ph=0, nn=NN, fs=FS)      # media onda trifásico
x_rtoc = sig.rtoc(vmax=311, ff=FF, ph=0, nn=NN, fs=FS)      # onda completa trifásico
# np.pi / 6  np.pi / 9
fig, axs = plt.subplots(4, 1, figsize=(12, 10))
axs[0].plot(tt, x_rmmo)
axs[0].set_title("Rectificador media onda")
axs[1].plot(tt, x_rmoc)
axs[1].set_title("Rectificador onda completa")
axs[2].plot(tt, x_rtmo)
axs[2].set_ylim(0, 311)
axs[2].set_title("Rectificador trifásico media onda")
axs[3].plot(tt, x_rtoc)
axs[3].set_ylim(0, 600)
axs[3].set_title("Rectificador trifásico onda completa")
for ax in axs.flat:
    ax.set_xlabel("t [s]")
    ax.grid(True)
 
fig.tight_layout()
plt.show()
```

# Rectificadores Controlados

```python
x_rmmo = sig.rmmo(vmax=311, ff=FF, ph=0, alpha=np.pi / 4, nn=NN, fs=FS)      # media onda monofásico
x_rmoc = sig.rmoc(vmax=311, ff=FF, ph=0, alpha=np.pi / 3, nn=NN, fs=FS)      # onda completa monofásico
x_rtmo = sig.rtmo(vmax=311, ff=FF, ph=0, alpha=np.pi / 6, nn=NN, fs=FS)      # media onda trifásico
x_rtoc = sig.rtoc(vmax=311, ff=FF, ph=0, alpha=np.pi / 9, nn=NN, fs=FS)      # onda completa trifásico
# np.pi / 6  np.pi / 9
fig, axs = plt.subplots(4, 1, figsize=(12, 10))
axs[0].plot(tt, x_rmmo)
axs[0].set_title("Rectificador media onda (alpha=pi/4)")
axs[1].plot(tt, x_rmoc)
axs[1].set_title("Rectificador onda completa (alpha=pi/3)")
axs[2].plot(tt, x_rtmo)
axs[2].set_ylim(0, 311)
axs[2].set_title("Rectificador trifásico media onda (alpha=pi/6)")
axs[3].plot(tt, x_rtoc)
axs[3].set_ylim(0, 600)
axs[3].set_title("Rectificador trifásico onda completa (alpha=pi/9)")
for ax in axs.flat:
    ax.set_xlabel("t [s]")
    ax.grid(True)
 
fig.tight_layout()
plt.show()
```

## Ruido

```python
x_noise = sig.noise_generator(var=0.05, nn=NN, fs=FS)
x_noisy = sig.noisy_sin(vmax=1, dc=0, ff=FF, ph=0, nn=NN, fs=FS, snr=15)

fig, axs = plt.subplots(2, 1, figsize=(12, 10))
axs[0].plot(tt, x_noisy)
axs[0].set_title("Señal sinusoidal con ruido (SNR=15 dB)")
axs[1].plot(tt, x_noise)
axs[1].set_title("Ruido blanco gaussiano (var=0.05)")
for ax in axs.flat:
    ax.set_xlabel("t [s]")
    ax.grid(True)
fig.tight_layout()
plt.show()
```

## Delta de Kronecker/Dirac

```python
delta = sig.kronecker_delta(NN)

plt.plot(tt, delta)
plt.xlabel("t [s]")
plt.ylabel("δ(t)")
plt.title("Delta de Kronecker")
plt.grid(True)
plt.show()
```
