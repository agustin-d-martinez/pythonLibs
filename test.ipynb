```python
from pathlib import Path
import sys

notebook_dir = Path().resolve()

target_dir = notebook_dir / "solver"
if str(target_dir) not in sys.path:
    sys.path.insert(0, str(target_dir))

from solver.block import *
from solver.junction import *
from solver.node import *
from solver.network import *
```

```python
a = Block(name='A')
b = Block(name='B')
c = Block(name='C')
d = Block(name='D')
e = Block(name='E')

j1 = Junction(name='J1')
j2 = Junction(name='J2')

red = Network()
red.add_block(a)
red.add_block(b)
red.add_block(c)
red.add_block(d)
red.add_block(e)
red.add_junction(j1)
red.add_junction(j2)

for block in [a, b, c,d,e]:
    block.add_terminal(name='V1')

c.add_terminal(name='V2')
e.add_terminal(name='V2')

red.connect(a.terminals['V1'], j1)
red.connect(b.terminals['V1'], j1)
red.connect(c.terminals['V1'], j1)
red.connect(d.terminals['V1'], j2)
#red.connect(j1, j2)
red.connect(e.terminals['V1'], j2)

nodes = red.build_nodes()

print("Nodes in the network:")
for node in nodes:
    print(node.physical_terminals)
```

```python
r1 = ZParamPortNetwork('R1', [[10]])
r2 = Resistance('R2', 90)
v1 = VoltageSource('Vg', 1)
ground = Ground('')

red = Network()
red.add(r1)
red.add(r2)
red.add(v1)
red.add(ground)

red.connect(v1.terminals['1'], r1.terminals['p0_in'])
red.connect(r1.terminals['p0_out'], r2.terminals['1'])
red.connect(v1.terminals['2'], r2.terminals['2'])
red.connect(ground.terminals['1'], r2.terminals['2'])

x, block_idx, node_idx = red.solve()

ans = block_idx[r1.terminals['p0_out']]

print(f"Corriente: {x[ans.i]}\nVoltaje: {x[ans.v]}")
```
